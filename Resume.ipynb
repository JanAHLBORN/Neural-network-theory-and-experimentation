# Theoretical neuroscience summary

## Biological neurons

### Morphology

Biological neurons consist of:
- a cell body, the soma. The soma is electrically charged in comparison to its surroundings and the electrical potential can be changed by input signals,
- dendrites, which connect the soma with the presynaptic neurons,
- the axon, which contains the output signal of the neuron, given to some postsynaptic neurons,
- Synapses, which transform electrical signals using chemical reactions.
Neurons communicate via short electrical pulses, called spikes.



### Excitatory and inhibitory nerve cells
There are different types of neuron cells (morphologically) that also have different methods of processing. In general, we differentiate between **excitatory** and **inhibitory** nerve cells:
- excitatory neurons lead to more activity and inhibitory reduce the activity,
- inhibitory nerve cells are said to prevent the activity to diverge,
- increasing the input to inhibitory neurons can decrease I and E activity.

### Ion channels and transition probabilities
The Cell membrane contains a variety of different proteins, and especially ion channels. Ion channels allow specific ions to pass the pore. They select ions via different technics:
- The pore size,
- Activation type of the channel (voltage gated, ligand gated (using a chemical messenger), ligand + voltage gated, via mechanical stress or second messengers).



The activity of ion channels fluctuate between different states. In general, we are interested in the stochastic properties of these states. Therefore, one can model ion channels as Markov processes, where the future state only depends on the current state and not on past ones. Consider two states $C$ and $O$, with the transition probabilites $P(C \rightarrow O) = \alpha$ and $P(O \rightarrow C) = \beta$. We now assume that the change of state only happens over time steps of $\Delta t$. Then we have:
- Probability to go to $O$ if the channel is in $C$:  $ \,\,\, P(O,t+\Delta t | C,t) = \alpha \Delta t$
- Probability to stay in $C$ if the channel is in $C$:   $\, P(C,t+\Delta t | C,t) = 1 - \alpha \Delta t$
- Probability to go to $C$ if the channel is in $O$:     $ \,\,\, P(C,t+\Delta t | O,t) = \beta \Delta t$
- Probability to stay in $O$ if the channel is in $O$:   $\, P(O,t+\Delta t | O,t) = 1 - \beta \Delta t$
- Probability to be in state $C$ at the time $t+\Delta t$ is: $P(C, t+\Delta t) = P(C,t+\Delta t | C,t)P(C,t) + P(C,t+\Delta t | O,t) P(O,t)$
- Probability is conserved: $P(C,t) + P(O,t) = 1$.

From the last equation one can derive the temporal evolution of the probability:
$$
\frac{\mathrm{d} P(C,t)}{\mathrm{d}t} = -\alpha P(C,t) + \beta P(O,t), \qquad \frac{\mathrm{d} P(O,t)}{\mathrm{d}t} = \alpha P(C,t) - \beta P(O,t),
$$
Using the conservation of probabilty one finds the **master equation**:
$$
\frac{\mathrm{d} P(C,t)}{\mathrm{d}t} = -(\alpha + \beta) P(C,t) + \beta 
$$


# The resting membrane potential

## The Nernst equation
We assume, that there is only one ion type that can pass the cell (let's say $K^+$ as in glial cells at rest). Other ions generate negative membrane potential and $K^+$ ions are attracted to the cell's inside and as a consequence, there is an ion flux into the cell. Now there is:
- a concentration gradient (inside the cell)
- net diffusion of ions out of the cell

At some potential these fluxes cancel each other leading to an equilibrium state. How are the potential and the concentration differences at equilibrium related to each other?
- Equilibrium thermodynamics: Boltzmann distribution  
$$
P(E) = \frac{1}{z} \mathrm{e}^{-E/K_B T}, \qquad z = \sum_E \mathrm{e}^{-E/K_B T} \quad \text{s.t.} \quad  \sum_E P(E) = 1
$$
- Let $V$ be the potential and $q$ the charge of the ion. The probability for an Ion to be inside the cell is $P(\mathrm{inside}) = \frac{1}{z} \mathrm{e}^{-qV/K_B T}$. Also the probability to be outside is $P(\mathrm{outside}) = 1/z$ (where the potential is assumed to be 0).
- If we consider many ions, the probability to be inside is just given by the fraction of ions being inside the cell. Also if we assume that the volumes inside and outside are equal, we get:
$$
\frac{[\mathrm{inside}]}{[\mathrm{outside}]} \approx \frac{P(\mathrm{inside})}{P(\mathrm{outside})} = \mathrm{e}^{-qV/K_B T}       
$$
where $[\mathrm{inside}]$ is the concentration.

Eventually, we get the Nernst eqation:
$$
V = - \frac{K_B T}{q} \, \ln\left( \frac{[\mathrm{inside}]}{[\mathrm{outside}]} \right)
$$

## The resting membrane potential for multiple ion types

The potential at which the net flux of a particular neuron type is 0, is called the reversal potential. Because the reversal potentials for the different ion types differs, there is no equilibrium. An electrical equilibrium where the transmembrane currents cancel, will still establish.

To compute the equilibrium (resting) membrane potential, we assume a linear current-voltage relationship:
$$
I_\mathrm{K}(t) = g_\mathrm{K} (V-E_\mathrm{K})
$$
where $g_K$ is something like the inverse of a resistance. Also we have the capacitor equation:
$$
I_C = \dot{Q} = C \dot{V}
$$
The total current flowing outside is then:
\begin{align*}
I &= C \dot{V} + I_{\mathrm{Na}} + I_{\mathrm{K}} + I_{\mathrm{Ca}} + I_{\mathrm{Cl}}\\
    &= C \dot{V} + g_\mathrm{Na} (V-E_\mathrm{Na}) + g_\mathrm{K} (V-E_\mathrm{K}) + g_\mathrm{Ca} (V-E_\mathrm{Ca})+ g_\mathrm{Cl} (V-E_\mathrm{Cl})
\end{align*}
At resting membrane potential, we have $I=0$ and $\dot{V}=0$ yielding:
$$
V_\mathrm{rest} = \frac{g_\mathrm{Na} E_\mathrm{Na} + g_\mathrm{Ca} E_\mathrm{
    Ca} + g_\mathrm{K} E_\mathrm{K} + g_\mathrm{Cl} E_\mathrm{Cl}}{g_\mathrm{Na} + g_\mathrm{K} +g_\mathrm{Ca} + g_\mathrm{Cl}}
$$

## Generation of spikes
**Potassium channel**: four identical gates, finding one open: probability: $n(t)$.

So the probability of finding a $K$-Channel open: $P_K(O,t) = n(t)^4$. 

Each subunit behaves like our two-state channel model from session 1, where we found the master equation. Thus we have:
\begin{align*}
\frac{\mathrm{d}n(t)}{\mathrm{d}t} &= \alpha_n(V) (1-n(t)) - \beta_n(V)n(t) \\
    &= -(\alpha_n(V) + \beta_n(V))n(t) + \alpha_n(V)
\end{align*}
It follows:
\begin{align*}
\frac{1}{\alpha_n(V) + \beta_n(V)} \frac{\mathrm{d}n(t)}{\mathrm{d}t} &=  -n(t) + \frac{\alpha_n(V)}{\alpha_n(V) + \beta_n(V)} \\
    &= -n(t) + n_\infty(V)
\end{align*}
If we define the time constant $\tau_n(V) = \frac{1}{\alpha_n(V) + \beta_n(V)}$, we find the exponential solution:
$$
n(t) = \mathrm{e}^{-t/\tau_n(V)} n(0) + (1-\mathrm{e}^{-t/\tau_n(V)}) n_\infty(V)
$$


**Sodium channel:** three activation subunits, one inactivation subunit

So the probability of finding an open channel: $P_\mathrm{Na}(O,t) = m(t)^3h(t)$. 

As before one can write down the master equation for $m$ and $h$. Eventually, there is also a leak current (unspecific) that has to be taken into account. We get the **HH equation:**
$$
C \dot{V}(t) = -\bar{g}_\mathrm{Na} m(t)^3 h(t) [V(t) - E_\mathrm{Na}] -\bar{g}_\mathrm{K} n(t)^4 [V(t) - E_\mathrm{K}] -g_\mathrm{leak} [V(t) - E_\mathrm{leak}] + I(t)
$$

## 3.2 A two-dimensional model, the $I_{Na,p} + I_K$-Model
In order to reduce the HH quation, we first set $m(t) = m_\infty (V(t))$, since this gate has a very small time constant. Also we need a spike upstroke and a downstroke:
- Upstroke: opening of the $\mathrm{Na}$ channels
- Downstroke: inactivation of $\mathrm{Na}$ channels; opening of $\mathrm{K}$ channels; leak channels.

It is totally sufficient if we only conside the opening of the $\mathrm{K}$ channels in addition with the leak channels. This gives
- $h(t) = 1$
- $\mathrm{Na}$ current becomes presistent (does not inactivate)
- the name of the model: persistent sodium plus potassium current model ($I_{Na,p} + I_K$-Model).

In therms of an equation, this can be expressed by:
$$
C \dot{V}(t) = - \bar{g}_\mathrm{Na} m_\infty(V(t))\, \left( V(t) - E_\mathrm{Na} \right) - \bar{g}_\mathrm{K} n(t)\, \left( V(t) - E_\mathrm{K} \right) - \bar{g}_L m_\infty(V(t))\, \left( V(t) - E_L \right) + I(t)
$$
also the time constant is $\tau_n \dot{n}(t) = -n(t) + n_\infty(V(t))$ and for further simplification we set $\tau_n(V(t)) = 1$. Simplified functional forms of $m_\infty, n_\infty$ are:
$$
m_\infty(V) = \frac{1}{1+ \mathrm{e}^{-(V-V_{1/2,m})/k_m}} , \quad n_\infty(V) = \frac{1}{1+ \mathrm{e}^{-(V-V_{1/2,n})/k_n}}
$$
with the slopes: $m_\infty(V_{1/2,m})/k_m) = 1/2$, $m'_\infty(V) = 1/4k$.

## 3.3 One-dimensional models

### 3.3.1 The leaky-integrate-and-fire (LIF) neuron
Assumption: above fixed threshold $\theta$, we get a spike and beneath we do not get any spike.

Action potential (=spike onset) is infinitely fast, so we do not explicitly model the spike (only throgh its impact on the postsynaptic neurons and the reseting of the membrane potential):
$$
V(t^*) \geq \theta \quad \Rightarrow \quad V(t^+) = V_\mathrm{reset}
$$
where $t^+$ is directly after $t^*$.
Further, one neglects all voltage dependent channels and just sticks with the leak current, yielding:
$$
C \dot{V}(t) = -g_L (V(t) - V_\mathrm{rest}) + I(t)
$$
This equation is analytically solvable, so this is efficient for mathematical analysis and event based (ves step. based) simulations.
We can rewrite the above equation:
$$
\tau_m \dot{V}(t) = - (V(t) - V_\mathrm{rest}) + R \, I(t)
$$
with the membrane time constant $\tau_m$ and the input resistance $R$. So the cell behaves like an RC-circuit and we have the dynamics of a low-pass filter with time constant $\tau_m$.

### 3.3.2 Threshold and reset in the quadratic integrate and fire model (QIF)

The QIF model is 1-dimensional and explicitly models the spike upstroke:
$$
C \dot{V}(t) = g_L \frac{(V(t) - \theta)(V(t) - V_\mathrm{rest})}{\theta - V_\mathrm{rest}} + I(t) 
$$
In this equation, we see that $\dot{V}(t) \propto V(t)^2$, like we can see in the beneath figure.



In the lower figure we see the vector field representation of the potential. In this representation trajectories cannot cross. Thus, we cannot get back from $V > \theta$ and introduce an artificial reset, to some potential smaller than $\theta$.

Connections to previous models:
\begin{align*}
C \dot{V}(t) &= g_L \frac{(V(t) -V_\mathrm{rest} + V_\mathrm{rest} - \theta)(V(t) - V_\mathrm{rest})}{\theta - V_\mathrm{rest}} + I(t)\\
             &= \frac{g_L}{\theta - V_\mathrm{rest}} (V(t) - V_\mathrm{rest})^2 + g_L \frac{V_\mathrm{rest} - \theta}{\theta - V_\mathrm{rest}} (V(t) - V_\mathrm{rest}) + I(t)\\
             &= a \, (V(t) - V_\mathrm{rest})^2 - g_L  (V(t) - V_\mathrm{rest}) + I(t)
\end{align*}
where $a \, (V(t) - V_\mathrm{rest})^2$ is the active part, that leads to spike generation and $- g_L  (V(t) - V_\mathrm{rest})$ corresponds to the leakage current. This equation is analytically solvable in many cases.
First, consider the case of $I(t)=0$. We:
- First further simplify the differential equation
- Measure time in multiples of $\tau_m = C/g_L$.
- Shift the voltage by $-V_\mathrm{rest}$, s.t. resting potential gets zero
- Measure voltage in multiples of $\theta$, s.t. threshold becomes 1.

With this notation, the ODE reduces to:
$$
\dot{V}(t) = (V(t) - 1) \, V(t)
$$
This ODE can be solved by 
$$
V(t) = \frac{V(0)}{V(0) - (V(0) - 1)\mathrm{e}^{t}}
$$
Let's discuss the result:
- At $t=0$: we simply get $V(0) = 1 > 0$.
- For $V(0)<1$: The denominator is always positive
- For $V(0)>1$: For $t\rightarrow \infty$, the denominator gets negative. That means we have a zero crossing (at finite time). This feature is called "Blow up of the solution". 

The denominator gets zero when:
$$
t_\infty = \mathrm{ln}\left( \frac{V(0)}{V(0) -  1} \right)
$$
for $V(0)>1$. So we can use $t_\infty$ as spike time. We can then reset to $-\infty$, when the potential reaches $\infty$.

### 3.3.3 The exponential integrate and fire model (EIF)

Instead of the quadratic upstroke we now have an exponential one:
$$
C \dot{V}(t) = g_L \Delta \,\mathrm{e}^{\frac{V(t) - \theta}{\Delta}} - g_L(V(t) - V_\mathrm{rest}) + I(t)
$$
where $\Delta$ is a tunable sharpness parameter. This model fits well the experimental neuron dynamics, after adding small modifications. But there are no analytical solutions for this differential equation (also not event based simulation, every time step needs to be simulated).



## 3.4 Zero dimensional point neurons
### 3.4.1 Binary neurons in discrete time

No intrinsic dynamics. Assume value 0 (inactive) or 1 (active, spiking) in each step depending in the input. The state $S(t)$ is given by the input current $I(t)$, given as argument to the heaviside function $\Theta$:
$$
S(t) = \Theta(I(t) - \theta)
$$
Straightforward modification: replace $\Theta \rightarrow g$, where $g$ is a continuous activation function. So, we more generally get:
$$
S(t) = g(I(t) - \theta)
$$
where now $S(t) \in \mathbb{R}$. This equation is more used in machine learning. In neuroscience, it is used to model the population of neurons.

## 4.2 Cable theory
### 4.2.1 The general cable equation

The cable equation is a partial differential equation for the voltage along a simplified neurite (dendrite or axon):
$$
c \frac{\partial V(x,t)}{\partial t} = \frac{a}{2 \tau_\mathrm{
long}} \frac{\partial^2 V(x,t)}{\partial x^2} - i_\mathrm{mem}(x,t) + i_\mathrm{e}(x,t)
$$
with: 
- $c$: specific capacitance (capacitance per unit area)
- $a$: diameter of the neurite
- $\tau_\mathrm{long}$: specific longitudinal resistance
- $i_\mathrm{mem}$: transmembrane current per surface area
- $i_\mathrm{e}$: injected current per surface area

In a passive cable we have:
$$
i_\mathrm{mem}(x,t) = \bar{g}_L (V(x,t) - V_\mathrm{rest})
$$
with $\bar{g}_L$ the conductance per surface area. Shifitng again the voltage, s.t. $V_\mathrm{rest} = 0$, gives the passiv ecable equation:
$$
c \frac{\mathrm{d} V(x,t)}{\mathrm{d} t} = \frac{a}{2 r_\mathrm{
long}} \frac{\partial^2 V(x,t)}{\partial x^2} - \bar{g}_L V(x,t) + i_\mathrm{e}(x,t)
$$

### 4.2.2 Derivation of the cable equation



So the idea is to seperate the cable into these compartements of length $\Delta x$ and then let $\Delta x \rightarrow \infty$. In this limit the differential equation for the compartement becomes a partial differential equation for the entire cable (cable equation).


We begin with:
$$
C \frac{\partial V(x,t)}{\partial t} = - I_\mathrm{mem}(x,t) + I_\mathrm{long}(x,t) - I_\mathrm{long}(x+\Delta x,t) + I_e(x,t)
$$
and use:
- $R_\mathrm{long} I_\mathrm{long}(x,t) = V(x,t) - V(x+\Delta x,t)$. 
- The capacity equation $C = 2\pi a \cdot \Delta x \cdot c$, with the radius of the compartement $a$.
- $I_\mathrm{mem}(x,t) = 2 \pi a \Delta x \, i_\mathrm{mem}(x,t)$.
- $I_e(x,t) = 2 \pi a \Delta x \, i_e(x,t)$.
- $R_\mathrm{long} = \frac{\Delta x}{\pi a^2} r_\mathrm{long}$ and thus we can write $I_\mathrm{long}(x,t) = \frac{\pi a^2}{\delta x} (V(x,t) - V(x+\Delta x,t))$.


And putting all this together yields:
\begin{align*}
&2 \pi a \Delta x c \frac{\partial V (x,t)}{\partial t} = - 2 \pi a \Delta x \, i_\mathrm{mem}(x,t) + I_\mathrm{long}(x,t) - I_\mathrm{long}(x+\Delta x,t) + 2 \pi a \Delta x i_e(x,t) \\
\Rightarrow & \frac{\partial V (x,t)}{\partial t} = - i_\mathrm{mem}(x,t)  + i_e(x,t) + \frac{I_\mathrm{long}(x,t) - I_\mathrm{long}(x+\Delta x,t)}{2 \pi a \Delta x c}
\end{align*}
In the limit of $\Delta x \rightarrow \infty$, we then get:
$$
\frac{\partial V (x,t)}{\partial t} = - i_\mathrm{mem}(x,t)  + i_e(x,t) + - \frac{1}{{2 \pi a \Delta x c}} \frac{\partial I_\mathrm{long}(x,t)}{\partial x}
$$
Using that for $\Delta x \rightarrow \infty$, we then geht:
$$I_\mathrm{long}(x,t) = - \frac{\pi a^2}{r_\mathrm{long}} \frac{\partial V (x,t)}{\partial x}$$
leads to:
$$
c \frac{\partial V(x,t)}{\partial t} = \frac{a}{2 r_\mathrm{long}} \frac{\partial^2 V(x,t)}{\partial x^2} - i_\mathrm{mem} + i_e(x,t)
$$
This is a diffudion equation with $D=a/2 c r_\mathrm{long}$ and particle (voltage) loss or gain.


### 4.2.3 The pasiv ecable equation

We have 
$$
i_\mathrm{mem}(x,t) = \bar{g}_L (V(x,t) - V_\mathrm{rest})
$$
and shift the voltage s.t. $V_\mathrm{rest}=0$ and introduce:
the electronic length (length at which the voltage decayed to its 1/e-th):
$$
\lambda = \sqrt{\frac{a}{2 r_\mathrm{long} \bar{g}_L}}.
$$
Putting this all toghether yields:
$$
\tau_m \frac{\partial V(x,t)}{\partial t} = \lambda^2 \frac{\partial^2 V(x,t)}{\partial x^2} - V(x,t) + \frac{1}{\bar{g}_L} i_e(x,t)
$$
with $\tau_m = C/\bar{g}_L$.

Most prevalent synapse in the brain are glutamatergic excitatory sinapses (the neuro transmitter they use is glutamate). Most frequent frequent receptors are AMPA and NMDA ion cahnnels.

Let's diskuss what the different parts of electric current of the AMPA channel mean:
$$
I_\mathrm{syn,e}(t) = g_\mathrm{syn,e}(t) (V(t) - E_\mathrm{syn,e})
$$
we can see that the conductance determines the slope and $E_\mathrm{syn,e}$ ($\approx 0$ for AMPA) the zero of the linear dependency of the current in the voltage.


The NMDA channel conduct $\mathrm{Na}^+$, $\mathrm{Ca}^{2+}$ and $\mathrm{K}^+$ ions. Synaptic transmitters are glutamate, but further conditions have to be fullfilled in order to open the channel. Most importantly, the postsynaptic sell needs to be depolarized. Otherwise, $\mathrm{Mg}^{2+}$ blocks the channel. Just like before, we would expect something like:

$$

## 6.2 Spike train models

### 6.2.1 Poisson process

Simplest model for a spike train with high irregularity and randomness. Spikes occur at random, with a rate $r(t)$. The probability to generate a spike in a short time interval $[t,t+\Delta t]$ is:
$$
P(t, t+\Delta t) = r(t) \Delta t \ll 1
$$
The probability to generate 2 spikes in a sufficiently short interval is negligible ($\ll 1$). So this is a typical Poisson process.

### 6.2.2 Some characteristics of the homogeneous Poisson process

*Definition:* In a homogeneous Poisson process the rate is constant: $r(t)=r$.

One major advantage of this process is that important features can be computed analytically especially the distribution of number of generated spikes in an interval $T$ and the distribution of the times passing between subsequent spikes (interspike interval ISI distribution).

**Computation of the number of spikes in $T$:** We assume that $T$ is binned into $T/\Delta t$ bins of length $\Delta t$. The probability of finding spikes in the bins $[t_1, t_1+\Delta t], [t_2, t_2+\Delta t], \dots, [t_n, t_n+\Delta t]$ is given by:
$$
p = \Pi_{i=1}^n P(t_i, t_i+\Delta t) = (r \Delta t)^n
$$
The probability that no spikes are present in the remaining bins is:
$$
\bar{p} = \Pi_{j \notin I}^n \left(1 - P(t_j, t_j+\Delta t)\right) = (1-r\Delta t)^{\frac{T}{\Delta t}-n}
$$
with $I=\{1,\dots, n\}$. So taken together, the probability $p_s$ to find all spikes exactly in the $n$ considered bins is:
$$
p_s = p \, \bar{p} = (r \Delta t)^n \, (1-r\Delta t)^{\frac{T}{\Delta t}-n}
$$

The binomial coefficient 
$
\begin{pmatrix}
    T/\Delta t \\
    n
\end{pmatrix}
$
gives the number of possibilities that the $n$ considered bins can be distributed among the $T/\Delta t$ bins of the trial. So the probability to find $n$ spikes in a (binned) trial of length $T$ is:
$$
p(n) = (r \Delta t)^n \, (1-r\Delta t)^{\frac{T}{\Delta t}-n}
\begin{pmatrix}
    T/\Delta t \\
    n
\end{pmatrix}
$$

We now want to let $\Delta t$ go to zero in order to find the probability for continuous space. First, we note that For small $\Delta t$ the relation:
$$
\mathrm{e}^{-r \Delta t} \approx 1-r\Delta t
$$
holds. Thus, we can write:
\begin{align*}
    (1-r\Delta t)^{\frac{T}{\Delta t}-n} \approx \mathrm{e}^{-r \Delta t}.
\end{align*}
Note that in simulation $r \Delta t$. Further, for small $\Delta t$ we have:
$$
    \begin{pmatrix}
        T/\Delta t \\
        n
    \end{pmatrix}  =\frac{\frac{T}{\Delta t} \left( \frac{T}{\Delta t} -1 \right) \dots \frac{T}{\Delta t} \left( \frac{T}{\Delta t} -n +1 \right)}{\frac{T}{\Delta t} \left( \frac{T}{\Delta t} -n \right)! n!}\approx \frac{\frac{T}{\Delta t}\left( {\Delta t} \right)^n}{n!}
$$
And we eventually get the probability:
$$
p(n) = \mathrm{e}^{-rT} \frac{(rT)^n}{n!}
$$
in the limit of small timesteps. This is a Poisson distribution with parameter $\lambda = rT$. To highlight the interpretation of $n$(number of spikes in $[0,T)$), we often write $P(n(0,T))$.

*Note:* 
- The Poisson distribution is a discrete distribution and $n$ is the variable, whereas $\lambda=rT = n(0,T)$ is a parameter. 
- Also, $\braket{n(0,T)} = \mathrm{lim}_{M\rightarrow \infty} \frac{1}{M} \sum_{m=1}^M n_m(0,T)$, where $n_m(0,T)$ is the mean number of spikes observed in the $m$-th trial.
- For simulations, one can rather distribute spikes, by drawing from a Poisson distribution or one can draw the number of spikes from a Poisson distribution and unifromly distribute the on the interval.
- **For an inhomogeneous Poisson process**: $\braket{n(0,T)} = \int_0^T r(t) \, \mathrm{d}t$



**Computation of the ISI distirbution:**
shift time origin s.t. the last spike was at zero. When is the next spike? Let's call the time of the next spike $\tau$. The probability to have no spike in $\tau$ times the probability to have a spike in $[T, T +\Delta t)$, gives the probability to have an ISI in $[T, T + \Delta t)$:
$$
P\left(\tau \in [T, T + \Delta t]\right) = \mathrm{e}^{-rT} r \Delta t
$$
The probability density of the ISI is therefore exponentially distributed:
$$
p\left(\mathrm{ISI}=\tau\right)=r\mathrm{e}^{-r \tau} 
$$

*Note:* 
- for simulations we can draw the ISI from the exponential distribution. Starting at zero, this then gives the first spike time, then we draw the next ISI from the distribution again and so on. 
- $T$ indicites the length of an interval we are examining and does not explicitly mean the spike train length.

More generally, the waiting time until the $(n+1)$-th spike is distributed according to:
$$
p_n(t_{n+1}=T) = \frac{r^{n+1}}{n!} T^n \,\mathrm{e}^{-rT}
$$
For $n=0$, we wait until the first spike (yielding the ISI distribution). $p_n(t_{n+1}=T)$ is Gamma distributed with shape parameter $\alpha = n+1$ and rate parameter $\beta=r$.

## 6.3 Characterizing spike trains

### 6.3.1 The spike rate

Concept of spike rate roots in the view that a measured spike train behaves similar to a Poisson spike train. There should be a temporal probability density $r(t)$ to observe a spike at a time $t$. This neglects relations between spikes. 

How can we estimate this rate? 

**1)** If we have only a single trial, we can compute a temporal average to estimate the temporal average
$$
\bar{r} = \frac{1}{T} \int_0^T r(t) \, \mathrm{d}t.
$$
The estimated average is then
$$
r_\mathrm{est} = \frac{n(0,T)}{T} \approx \frac{\int_0^T r(t) \, \mathrm{d}t}{T} = \bar{r}
$$
(we approximated $n(0,T)$ by its ensemble average over trials, $\braket{n(0,T)} = \int_0^T r(t) \, \mathrm{d}t$). The approximation gets better the larger $T$ is, because of the law of large numbers.

**2)** Another approach: Spike train smoothing:
$$
S = \sum_{t_{sp}} \delta(t-t_{sp})
$$
Spikes could also have appeared earlier or later:
$$
r_\mathrm{rest}(t) = \sum_{t_{sp}} K(t-t_{sp})
$$
with $K$ the window function, smoothing kernel, for example Gaussian. We can write $r_\mathrm{rest}(t)$ as a convolution between $K$ and the spike train:
\begin{align*}
    r_\mathrm{rest}(t) &= \sum_{t_{sp}} K(t-t_{sp}) \\
    &= \sum_{t_{sp}}  \int_{-\infty}^\infty K(\tau) \delta(\tau - (t-t_{sp})) \, \mathrm{d}\tau \\
    &= \int_{-\infty}^\infty \sum_{t_{sp}} K(\tau) \delta(\tau - (t-t_{sp})) \, \mathrm{d}\tau \\
    &= \int_{-\infty}^\infty K(\tau) S(t-\tau) \, \mathrm{d}\tau \\
    &= (K*S)(t)
\end{align*}
If we have a larger number of repetitions $M$ of a trial, then we can compute the spake rate by trial averaging:
\begin{align*}
    r_\mathrm{rest}(t) &= \frac1M \sum_{m=1}^M \frac{n_m(t, t+\Delta t)}{\Delta t} \\
    &\approx \lim_{M \rightarrow \infty} \frac1M \sum_{m=1}^M \frac{n_m(t, t+\Delta t)}{\Delta t}\\
    &= \frac{\braket{n(t,t+\Delta t)}}{\Delta t} \\
    &\approx \frac{r(t) \Delta t}{\Delta t} \\
    &= r(t)
\end{align*}

**3)** If we have a larger number of $N$ neurons, we can do the same: compute a population average, i.e. average over an ensemble of neurons. This, especially if the neurons behave similarly (have the same underlying spike rate $r(t)$).
$$
r_\mathrm{rest}(t) = \frac1N \sum_{m=1}^N \frac{n_m(t, t+\Delta t)}{\Delta t} \approx r(t)
$$

### 6.3.2 Spike number and Fano factor

Mean number of spikes $\braket{n(0,T)}$. We expect this to scale linearly with $T$. What about the fluctuations? If there is no long-term drift (changes in the spike train properties) and $T$ is long, we can split it into already long pieces $\tilde{T}$ with $T = M \tilde{T}$. Then, $n(m \tilde{T}, (m+1) \tilde{T})$ are basically independent random variables and behave like independent, alike trials.
- $n(0,T) = \sum_{m=0}^{M-1} n(m \tilde{T}, (m+1) \tilde{T})$
- $\braket{n(0,T)} = \sum_{m=0}^{M-1} \braket{n(m \tilde{T}, (m+1) \tilde{T})}$

So if we sum independent random variables their means (see above), but also their variances add up.
Also, the variance scales like $T$. Therefore, it makes sense to define the Fano factors as a characteristic for spike trains:
$$
FF = \frac{\mathrm{Var}(n(0,T))}{\braket{n(0,T)}}
$$

In Poisson processes, the number of spikes in an interval of length $T$ is Poisson distributed, with parameter $\lambda=\braket{n(0,T)}$. The mean number of spikes is thus $\braket{n(0,T)}$ and the variance is $\mathrm{Var}(n(0,T)) = \braket{n(0,T)}$. So the FF of a Poisson process is thus 1. This holds independent of the rate $r(t)$ and its fluctuations. 
We can thus use the FF as a means to investigate how close the variability of a spike train is to that of a Poisson process.

### 6.3.3 Interspike interval distribution and the coefficient of variation

We have seen, that the ISI distribution is a useful characteristic of a spike train. A single characteristic for its variability is the coefficient of variation
$$
CV = \frac{\sigma(\mathrm{ISI})}{\mu(\mathrm{ISI})},
$$
where $\mu(\mathrm{ISI})$ is the mean ISI and $\sigma(\mathrm{ISI})$ is the standard deviation of the ISIs. It measures the width of the ISI distribution, the variability of the ISIs.

In the homogeneous Poisson process with rate $r$, the ISIs are distributed according to $p(\tau = \mathrm{ISI}) = r \mathrm{e}^{-r \tau}$. The mean and the standard deviation are thus 
$$
\mu(\mathrm{ISI}) = \frac1r , \qquad \sigma(\mathrm{ISI}) = \frac1r
$$
leading to $CV=1$, which is independent of the rate. Also, the $CV$ is used as a simple measure to see how close the variability of a spike train is to that of a Poisson process. Note that for:
- regular spike trains: $CV < 1$
- Poisson-like spike trains: $CV=1$
- irregular spike trains: $CV > 1$

### 6.3.4 Correlations 

We say that 
$$
r(t) \, \Delta t \approx P(t_{sp} \in [t + \Delta t)) = \braket{n(t,t+\Delta t)} = \lim_{M \rightarrow \infty} \frac1M \sum_{m=1}^M \int_t^{t+\Delta t} S_m(s) \mathrm{d}s = \langle \int_t^{t+\Delta t} S(s) \mathrm{d}s \rangle "=" \int_t^{t+\Delta t} \braket{S(s)} \mathrm{d}s  \approx \braket{S(t)} \Delta t
$$
where $S_m(s)$ denotes the spike train as a sum of delta functions. Also, we formally interchanged averaging and integration, because both are linear and assumed that $S(s)$ is sufficiently smooth. Actually, $\braket{S(t)} \Delta t$ can bee seen as an abbreviation for $\langle \int_t^{t+\Delta t} S(s) \mathrm{d}s \rangle$. Thus, we can see that
$$
r(t) = \braket{S(t)}
$$

**The autocorrelation:**

$r(t)$ gave the probability of occurrence of a spike in a short interval. Now, we can ask about the occurrence of one spike in one interval ($t_{sp} \in [t, t+\Delta t)$) and another spike in another interval ($\tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t)$):
$$
P(t_{sp} \in [t, t+\Delta t) \text{ and } \tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t)) = \int_t^{t+\Delta t} \int_{\tilde{t}}^{\tilde{t} + \Delta t} C(s, \tilde{s}) \mathrm{d}s \mathrm{d}\tilde{s} = C(t,\tilde{t}) \Delta t^2
$$
We can thus write
\begin{align*}
    P(t_{sp} \in [t, t+\Delta t) \text{ and } \tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t)) &= \lim_{M\rightarrow \infty} \frac1M \sum_{m=1}^M \int_t^{t+\Delta t} S_m(s) \mathrm{d}s \int_{\tilde{t}}^{\tilde{t}+\Delta t} S_m(\tilde{s}) \mathrm{d}\tilde{s}\\
    &= \braket{ \int_t^{t+\Delta t} S(s) \mathrm{d}s \int_{\tilde{t}}^{\tilde{t}+\Delta t} S(\tilde{s}) \mathrm{d}\tilde{s} }\\
    &"=" \int_t^{t+\Delta t} \mathrm{d}s \int_{\tilde{t}}^{\tilde{t}+\Delta t} \braket{S(s) S(\tilde{s})} \mathrm{d}\tilde{s} \\
    & \approx \braket{S(t) S(\tilde{t})} \Delta t^2
\end{align*}
So we eventually get: 
$$
C(t, \tilde{t}) = \braket{S(t)S(\tilde{t})}
$$
There is a very nice interpretation of $C(t,\tilde{t})$ in terms of conditional probability, more precisely conditional rates. The conditional rate $r(\tilde{t} | t_{sp}=t)$ tells the rate at some time $\tilde{t}$ given that we know that there was a spike at time $t$. 

The conditional probability to have a spike at $\tilde{t}$ given that we have a spike at $t$ is:
$$
P(\tilde{t}_{sp} \in [\tilde{t}, \tilde{t} + \Delta t] | t_{sp}=t) = r(\tilde{t} | t_{sp}=t) \Delta t
$$

Let's again consider the ISI distribution. The ISI tells s about the probability that two subsequent spikes have a certain temporal distance. We can formally write this as a conditional probability:
$$
P(\tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t) \text{ and no spike in } [t, \tilde{t}) \, | \, t_{sp}=t ) = p(ISI = \tilde{t} - t) \, \Delta t
$$
If we remove the restriction that $\tilde{sp}$ and $t_{sp}$ are subsequent, we get again:
$$
P(\tilde{t}_{sp} \in [\tilde{t}, \tilde{t} + \Delta t] | t_{sp}=t) = r(\tilde{t} | t_{sp}=t) \Delta t
$$
We now want to relate the conditional rate to the autocorrelation $C(t, \tilde{t}) = \braket{S(t) S(\tilde{t})}$.
The definition of the autocorrelation was:
\begin{align*}
    P(t_{sp} \in [t, t+\Delta t) \text{ and } \tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t)) &= \int_t^{t+\Delta t} \mathrm{d}s \int_{\tilde{t}}^{\tilde{t}+\Delta t} \braket{s_m(s) S_m(\tilde{s})} \mathrm{d}\tilde{s} \\
    & \approx C(t, \tilde{t}) \Delta t^2
\end{align*}
assuming that $C$ is sufficiently smooth at $t, \tilde{t}$. Thus, we can write:
\begin{align*}
    C(t, \tilde{t}) &= \frac{P(t_{sp} \in [t, t+\Delta t) \text{ and } \tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t))}{\Delta t^2} \\
    &= \frac{p(\tilde{t}_{sp} \in [\tilde{t}, \tilde{t}+\Delta t) \, | \, t_{sp} \in [t, t+\Delta t))}{\Delta t^2} \cdot p(t_{sp} \in [t, t + \Delta t)) \\
    &\approx \frac{p(\tilde{t}_{sp} \in [\tilde{t}, \tilde{t} + \Delta t) \, | \, t_{sp} = t)}{\Delta t} \cdot r(t) \\
    &= r(\tilde{t} | t_{sp} = t) \, r(t)
\end{align*}
The correlation function, the rate at which spikes occur at $t$ and $\tilde{t}$, is the conditional rate with which we have a spike at $\tilde{t}$, given that we have one at $t$ times the rate with which we have a spike at $t$.

*Note* that the conditional rate is not giving the probability for subsequent spikes (there could be spikes between $\tilde{t}$ and $t$).

Often, we are only interested in the probability that a spike occurs $\tau$ after another. We can then write $\tilde{t} = t + \tau$ and average $C(t, t + \tau)$ over $t$:
\begin{align*}
    C(\tau) &= \frac{1}{T-\tau} \int_0^{T-\tau} \mathrm{d}t \, C(t, t+ \tau) \\
    &= \frac{1}{T-\tau} \int_0^{T-\tau} \mathrm{d}t \, \braket{S(t) S(t+\tau)} \\
    &= \overline{\braket{S(t) S(t+\tau)}}^t 
\end{align*}

Note that we have two expectation vales here: an ensemble and a temporal average.

When analyzing data, if we have sufficiently trial and a (roughly) stationary spike train, we often do not need the ensemble average, since all the statistically relevant realizations of the spike train occur in the trial (**ergodicity**). On the other hand, we can omit the time average if we have a stationary process (stationarity means that there is no long time change in the system), because the ensemble average is no longer time dependent:
$$
\overline{\braket{S(t) S(t+\tau)}}^t  = \overline{\braket{S(0) S(0+\tau)}}^t = \braket{S(t) S(t+\tau)}
$$
But one cannot combine both "aussagen". Note that the time average needs to be taken over a suitable interval:
$$
    C(\tau) = \frac{1}{T-\tau} \int_0^{T-\tau} \mathrm{d}t \, \braket{S(t) S(t+\tau)}
$$
if $\tau > 0$ (which we focussed on) and
$$
   C(\tau) = \frac{1}{T+\tau} \int_{-\tau}^{T} \mathrm{d}t \, \braket{S(t) S(t+\tau)}
$$
if $\tau < 0$.


Also, for small enough $\Delta t$ we can write the probability to have somewhere in the spike train a spike pair with temporal difference within $[\tau, \tau + \Delta t)$ as:
$$
C(\tau) \cdot T \cdot \Delta t \approx C(\tau) \cdot (T-\tau) \cdot \Delta t
$$
If the underlying spike train is stationary, $C(t, \tilde{t})$ only depends on the temporal difference between $t$ and $\tilde{t}$, then 
$$
C(\tau) = C(t, t+ \tau)
$$
for all $t$. Then, 
$$
C(\tau) = r(\tau | t_{sp} = 0) \cdot r
$$


If spikes at $t$ and $t + \tau$ do not influence each other, we can take the ensemble average at each time individually (Poisson processes for example):
$$
\braket{S(t) S(t+\tau)} = \braket{S(t)}\braket{S(t+\tau)} = r(t) \, r(t+\tau)
$$
So correlations between spikes are reflected by an excess or reduced co-spiking compared to this values. Therefore, the autocorrelation is often further normalized, to the **autocovariance** (which is often also called autocorrelation):
$$
Q(\tau) = \overline{\braket{(s(t) - r(t))(S(t+\tau)-r(t+\tau))}}^t
$$
This is zero if spikes do not influence each other. If excess spiking at $t$ leads to excess spiking at $t+\tau$, $Q$ is positive (correlation). If excess spiking at $t$ leads to reduced spiking at $t+\tau$, $Q$ is negative (anticorrelation).

Often, also temporally averaged rates are subtracted, then this holds for stationary processes. And another normalization is
$$
Q_1(\tau) = \frac{C(\tau)}{\bar{r}^2}
$$
For stationary precesses, correlation between spikes are signaled by $Q_1(\tau) > 1$.


**Cross correlations:**

Estimates correlations between spike trains of different neurons, yielding
$$
C_{ij}(t, \tilde{t}) = \braket{S_i(t) S_j(\tilde{t})}, \qquad C_{ij}(\tau) = \overline{\braket{S_i(t) S_j(\tilde{t})}}^t
$$
and 
$$
Q_{ij}(\tau) =  \overline{\braket{(S_i(t)-r_i(t)) (S_j(\tilde{t}) - r_j(t))}}^t, \qquad Q_{1, ij}(\tau) = \frac{C_{ij}(\tau)}{\bar{r}_i \bar{r}_j} 
$$

For example, if the spikes of the neurons do not influence each other, $Q_{ij}(\tau) = 0$.

### Correlation properties and explicit values

$C(\tau)$ is symmetric in $\tau$ and there is a Dirac-delta-peak at $\tau=0$. The same holds for $Q$ and $Q_1$. We can split $C(\tau)$ into:
$$
C(\tau) = \bar{r}\delta(\tau) + C(\tau)
$$
where the bar means the temporal average.

*Example: Poisson process for $\tau > 0$ :*
\begin{align*}
    C(\tau) &= \overline{\braket{S(t) S(t+\tau)}}^t \\
    &= \overline{\braket{S(t)} \braket{S(t+\tau)}}^t \\
    &= \overline{r(t) r(t+\tau)}^t
\end{align*}
and if we insert the delta peak we get:
$$
    C(\tau) = \bar{r} \delta(t) + \overline{r(t) r(t+\tau)}^t
$$
(example end)


Let's now derive the symmetry. For stationary processes (no need for time average):
\begin{align*}
    C(\tau) = C(t,t+\tau) = C(s-\tau, s) = \braket{S(s-\tau) S(s)} = \braket{S(s) S(s-\tau)} = C(-\tau)
\end{align*}
For the non-stationary process we get:
\begin{align*}
    C(\tau) &= \frac{1}{T-\tau} \int_0^{T-\tau} \mathrm{d}t \, \braket{S(t) S(t+\tau)} \\
    &= \frac{1}{T-\tau} \int_\tau^{T} \mathrm{d}s \, \braket{S(s-\tau) S(s)} \\
    &= \frac{1}{T-\tau} \int_\tau^{T} \mathrm{d}t \, \braket{S(t) S(t-\tau)} \\
    &= C(-\tau)
\end{align*}

Where does the Dirac-delta distribution at the center comes from? It comes from the fact that whenever we have a spike at $t$ we also have a spike at $t+\tau$ with $\tau=0$, while for any other value of $\tau$, we only have a spike with some probability around $t+\tau$. In more detail:
\begin{align*}
    P(t_{sp} \in [t,t+\Delta t) \text{ and } \tilde{t}_{sp} \in [\tilde{t}, \tilde{t} + \Delta t)) = \int_t^{t+\Delta t} \mathrm{d}s \int_{\tilde{t}}^{\tilde{t}+\Delta t} \mathrm{d}\tilde{s} \, \braket{S(s)S(\tilde{s})}
\end{align*}
if now $t=\tilde{t}$, we get
\begin{align*}
    P(t_{sp} \in [t,t+\Delta t) \text{ and } \tilde{t}_{sp} \in [\tilde{t}, \tilde{t} + \Delta t)) = P(t_{sp} \in [t,t+\Delta t) \text{ and } \tilde{t}_{sp} \in [t, t + \Delta t)) ) = P(t_{sp} \in [t,t+\Delta t)) \approx r(t) \Delta t
\end{align*}
Therefore,
\begin{align*}
    \int_t^{t+\Delta t} \mathrm{d}s \int_{t}^{t+\Delta t} \mathrm{d}\tilde{s} \, \braket{S(s)S(\tilde{s})} \approx r(t) \Delta t
\end{align*}
This is the case if
\begin{align*}
    \braket{S(s)S(\tilde{s})} = r(s) \delta(s-\tilde{s}) + c(s,\tilde{s})
\end{align*}
with $c(S,\tilde{S})$ finite (if we plug this into the integral we use that the integral over delta is 1 and that the finite part is of order $\Delta t^2$). Thus, we can write:
$$
    C(\tau) = \bar{r} \delta(t) + c(\tau)
$$


The cross correlation does not have a delta peak at $\tau=0$ and it is not symmetric, but it satisfies:
$$
C_{ij}(\tau) = C_{ji}(-\tau)
$$


# 7. The balanced state

The occurrence of irregular spike trains was historically a conundrum (mistery). For self-consistency, we expect that each neuron which receives thousands of highly irregular, often only weakly correlated inputs must generate a highly irregular output. But will this happen in general? 

Consider the non-leaky (perfect) integrate-and-fire neuron:
$$
C \frac{\mathrm{d}V}{\mathrm{d}t} = I(t)
$$
Let us assume that the synapses are very simple. The synaptic currents have infinitesimal rise and decay, so they consist of the unfiltered spike trains:
$$
\frac{I(t)}{C} = \sum_{i=1}^N w_i \sum_k \delta(t-t_{sp,i,k})
$$
so there is no exponential decay in the output. The neuron spikes and is reset when $V(t)$ reaches or exceeds the threshold $\theta$:
$$
V(t) \geq \theta \quad \Rightarrow \quad V(t^+) = V_\mathrm{reset} = 0.
$$
Let's additionally assume that all inputs are excitatory, $w_i > 0$, and have same size $w_i = w$.

The number of input spikes that need to arrive to generate an output spike after the neuron has been reset is
$$
k = [\frac{\theta}{w}]
$$
where "[]" here is the Gauß braket: ceiling function that gives the smallest integer larger or equal to $\theta / w$. Further, we assume that the superposition of the input spike trains is a homogeneous Poisson process, which is often a good approximation. We know from lecture 9 that in a Poisson process the waiting time $T$ until we have the $k$-th spike arrival is gamma distributed
$$
p(t_k \in [T, T+\Delta t]) = \mathrm{e}^{-rT} \frac{(rT)^{k-1}}{(k-1)!}
 r\Delta t
$$
This directly provides the ISI distribution of the output neuron,
$$
    p(ISI = T) = \frac{r^k}{(k-1)!}\, T^{k-1} \mathrm{e}^{-rT}
$$

The mean and variance are 
$$
\mu(ISI) = \frac{k}{r}, \qquad \mathrm{Var}(ISI) = \frac{k}{r^2}
$$
So the CV is:
$$
CV = \frac{\sigma(ISI)}{\mu(ISI)} = \frac{1}{\sqrt{k}}, 
$$
independent of $r$. How large is $k$ for biological neurons?

Self-consistency argument: A neuron has $K$ presynaptic neurons (with $K$ in the thousands). Each one spikes at a certain rate $r_{pre}$. Self-consistency $r_{out} = r_{pre}$. Now the input spike train has $r = K \, r_{pre}$. Thus, we get:
$$
r_{out} = r_{pre} = \frac{r}{K} , \qquad K=k 
$$
Therefore, $K= [\frac{\theta}{w}] \approx \theta / w$ for $K \gg 1$ and thus $w$ must be small. And the CV is also small.
So the output spike trains become very regular and this is a general problem. So there must be inhibition that balances the excitation. Spikes should only be generated by fluctuations in the input. A spike is generated if there is by chance a period of more excitation and/or less inhibition.

Inhibition balances excitation leading to a balanced state.

# 8. Global descriptions of neuronal networks

## 8.1 Two types of rate networks

Rate networks model the interaction of neuron populations via spike rates, without modelling individual spikes. The state of a signal population (or for simplicity called neuron) $i$ is described by an overall activation level $x_i(t)$. At this activation level, the population generates a spike rate $r_i(t) = f(x_i(t))$, where $f$ is a generally nonlinear function. The strength of the impact of $r_j(t)$ on the activation of population $i$ is given by the coupling strength (weight) $w_{ij}$. Without input, activations decay exponentially to zero. Overall, the rate network dynamics are thus given by:
$$
\tau \, \frac{\mathrm{d}x_i(t)}{\mathrm{d}t} = -x_i(t) + \sum_{j=1}^N w_{ij} f(x_j(t)) + I_{i}(t)
$$
where $N$ is the number of populations (for large $N$ self interaction does not have much effect). $x$ is a low-pass filtered version of the input, like the subthreshold voltage of a LIF neuron. So, $x_i(t)$ may be interpreted as some kind of measure of the subthreshold voltages of the neurons in population $i$. Then, $\tau$ is the membrane time constant $\tau = \tau_m$.
Often $x_i$ is also interpreted as the input current to population $i$. Then, $\tau$ is the synaptic time constant $\tau = \tau_s$.
In matrix form, we have:
$$
\tau \,  \frac{\mathrm{d}\vec{x}(t)}{\mathrm{d}t} = -\vec{x}(t) + W \cdot f(\vec{x}) + \vec{I}(t)
$$
where $f$ acts element-wise on its argument vector. 



Often, one formulates an equation directly for the rates. For this, one assumes that the inputs to a population sum up linearly, like before, to $\sum_j w_{ij} r_j(t) + I_i(t)$. In the steady state, like before, we want to have 
$$
r_i = f \left(\sum_{j=1}^N w_{ij} r_j + I_i \right).
$$
But now we assume that $r_i$ exponentially approaches this state with time constant $\tau$. If $w_{ii}=0$, thus
$$
\tau \frac{\mathrm{d}r_i(t)}{\mathrm{d}t} = -r_i(t) + f  \left(\sum_{j=1}^N w_{ij} r_j + I_i \right)
$$
Further, one assumes that the exponential relaxation holds in general,
$$
\tau \frac{\mathrm{d}r_i(t)}{\mathrm{d}t} = - r_i(t) + f  \left(\sum_{j=1}^N w_{ij} r_j + I_i(t) \right)
$$
So, $r$ is in this second model type a low-pass filtered version of the non-linearly transformed total input. And before, $r$ was a non-linearly transformed version of the low-pass filtered total input. However, it can be shown that the two equations are equivalent, in the sense that they specify the same dynamical systems. One can show this, by transforming the variables $x(t)$ and $r(t)$ into each other (the transformation is not $\vec{r}(t) = f(\vec{x}(t)$). One can transform each solution $r(t)$ of the r-model equation into a solution $x(t)$ of the x-model equation,  after appropriately adapting its external input current vis versa.

## 8.2 oscillations

### 8.2.1 Oscillations in the brain

A hallmark of overall brain activity is that it is often oscillating. Such oscillations are measured in neuronal mass signal like the LFP (local field potential) and the EEG (electroencephalogram). 

### 8.2.2. Modelling oscillations

Several models exist. One famous, successful one is the two population rate model: populations $E$ (excitatory) and $I$ (inhibitory) acting on each other and on themselves. We consider a r-model
\begin{align*}
    \tau_E \frac{\mathrm{d}r_E}{\mathrm{d}t} &= -r_E(t) + \left[ w_{EE} r_E(t) + w_{EI} r_I(t) + I_E(t) \right]_+ \\
    \tau_I \frac{\mathrm{d}r_I}{\mathrm{d}t} &= -r_I(t) + \left[ w_{II} r_I(t) + w_{IE} r_E(t) + I_I(t) \right]_+ 
\end{align*}
with the rectification function
$$
f = [x]_+ =
    \begin{cases} 
            & x, \text{for } x \geq 0 \\
            & 0 , \text{else}
    \end{cases}
$$
Note that:
- $\tau_E \approx \tau_I$: Convergence to a stable fixed point, because $r_I$ can follow $r_E$ sufficiently quickly
- $\tau_E \ll \tau_I$: The system reaches constant oscillation for different initial values. The is a unstable fix point but a heteroclinic orbit.
